**Building Markov transition matrix** 

The possible states are the DNA bases: A, C, G, and T.
This function calculates how likely one nucleotide is to follow another.

Input:

sequence (string): DNA sequence to analyze.

Output:

transition_matrix: A nested dictionary showing transition probabilities.
Example: transition_matrix['A']['C'] gives the chance of C appearing after A.

```python
import pandas as pd

# Sample DNA string
dna_str = "CTAGTTCAGTTATTAATCTGGCTTATATACTGACTACTATATCTCTTATATATACTG"

# Define DNA bases
bases = ['A', 'C', 'G', 'T']

# Initialize nested dictionary to store base pair counts
base_pair_counts = {base: {b: 0 for b in bases} for base in bases}

# Tally the frequency of consecutive base pairs
for idx in range(len(dna_str) - 1):
    current_base = dna_str[idx]
    following_base = dna_str[idx + 1]
    if current_base in bases and following_base in bases:
        base_pair_counts[current_base][following_base] += 1
    else:
        print(f"Unexpected character '{current_base}' or '{following_base}' found. Ignored.")

# Create probability matrix from counts
prob_matrix = {b1: {} for b1 in bases}

for b1 in bases:
    total = sum(base_pair_counts[b1].values())
    if total > 0:
        for b2 in bases:
            prob_matrix[b1][b2] = base_pair_counts[b1][b2] / total
    else:
        for b2 in bases:
            prob_matrix[b1][b2] = 0.0

# Convert to DataFrame for display
transition_df = pd.DataFrame(prob_matrix)

print("Markov Transition Matrix: \nrows: current base\ncolumns: next base")
print(transition_df)
```

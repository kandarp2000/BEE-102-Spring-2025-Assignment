**Multi-line sequence fasta to single-line fasta file**

Parameters:
- input_file: Path to the input multi-line FASTA file.
- output_file: Path where the single-line FASTA file will be written.

```python
input_path = "multiline_input.fasta"
output_path = "output.fasta"

with open(input_path, 'r') as fasta_in, open(output_path, 'w') as fasta_out:
    
    seen_header = False

    for entry in fasta_in:
        # Remove trailing newline
        entry = entry.strip('\n')
        
        # If header line, manage spacing and write
        if entry.startswith('>'):
            if seen_header:
                fasta_out.write('\n')
            fasta_out.write(entry + '\n')
            seen_header = True
        else:
            # Write sequence without line breaks
            fasta_out.write(entry.strip())
    
    # Add newline at end
    fasta_out.write('\n')

print(f"Converted FASTA saved to '{output_path}'.")
```
